## Hook
程序在关键执行节点预留一个入口，当事件发生时，自动调用用户注册的函数。

流程：
1. 创建Hook对象
2. 注册Hook对象
3. 运行代码
4. 触发Hook对象

## 项目定位
所属架构：工具层

Hook在工具执行的生命周期上挂载自动化动作，是工具层和安全层的桥梁。

```python
@dataclass
class Hook:
    id: str
    event: str
    action: Action
    condition: ConditionGroup | None = None
    reject: bool = False
    once: bool = False
    async_exec: bool = False
    executed: bool = False
```

Hook 是整个系统的基本单元，一个 Hook 表示在某个事件发生时，如果满足条件，就执行某个动作

```python
@dataclass
class Action:
    type: str
    command: str = ""
    message: str = ""
    url: str = ""
    method: str = "POST"
    body: str = ""
    headers: dict[str, str] = field(default_factory=dict)
    prompt: str = ""
    timeout: int = 30
```
